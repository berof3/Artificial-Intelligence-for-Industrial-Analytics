2. Gradient Descent using PyTorch

```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# Define f(x) using torch math
def f(x):
    return -torch.exp(-(x - np.pi)**2) + 0.01 * (x - np.pi)**2

# Gradient Descent function with full explanation
def run_gd(eta):
    x = torch.tensor(0.0, requires_grad=True)   # Initial guess
    trajectory_x = []                           # store x-values for plotting
    trajectory_y = []                           # store f(x) values

    for step in range(1000):

        fx = f(x)                               # Compute function value

        # Convert tensors -> floats for plotting (as required in assignment)
        x_np = float(x.detach().numpy())
        trajectory_x.append(x_np)

        f_np = float(fx.detach().numpy())
        trajectory_y.append(f_np)

        fx.backward()                           # Compute gradient df/dx

        with torch.no_grad():                   # Disable gradient tracking
            x -= eta * x.grad                   # Gradient Descent update

        x.grad.zero_()                          # Reset gradient for next iteration

    return trajectory_x, trajectory_y, x         # Return path + final x

# Run all three step length factors (eta)
traj1_x, traj1_y, x1 = run_gd(1.0)
traj2_x, traj2_y, x2 = run_gd(0.1)
traj3_x, traj3_y, x3 = run_gd(0.01)

# Plot the result
plt.plot(traj1_x, traj1_y, label="η = 1.0")
plt.plot(traj2_x, traj2_y, label="η = 0.1")
plt.plot(traj3_x, traj3_y, label="η = 0.01")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Gradient Descent Trajectories")
plt.legend()
plt.grid(True)
plt.show()
```

Observation for different η:

**η = 0.1 - BEST**

- Converges smoothly
- Reaches the minimum near x ≈ 3.3
- Trajectory is stable

**η = 0.01 - VERY SLOW**

- Converges, but tiny steps
- Needs many iterations
- Trajectory barely moves at first

**η = 1.0 - FAIL**

- Steps are too large
- Jumps back and forth
- Often diverges to huge values
- Trajectory shows wild oscillations

Gradient descent needs a step size that is "big enough but not too big".

Exercise D.2: PyTorch Gradient Descent - two variables

```python
import numpy as np
import torch
import matplotlib.pyplot as plt

# Function for contour plotting (NumPy)
def f_float(x1, x2):
    return 4*(x1 - 3)**2 + (x2 - 2)**2

# Function for PyTorch gradient descent (PyTorch)
def f_torch(x):
    x1, x2 = x[0], x[1]
    return 4*(x1 - 3)**2 + (x2 - 2)**2

# Gradient Descent Parameters
x_start = np.array([-7.0, -8.0])
eta = 0.03  # best step size
n_max = 100

# Initialize starting point as PyTorch tensor
x = torch.tensor(x_start, requires_grad=True)

# Lists to store trajectory
trajectory_x1 = []
trajectory_x2 = []

# Compute the Gradient Descent
for n in range(n_max):
    fx = f_torch(x)

    # Record current position
    x_np = x.detach().numpy()
    trajectory_x1.append(x_np[0])
    trajectory_x2.append(x_np[1])

    # Compute gradient
    fx.backward()

    # Update x
    with torch.no_grad():
        x -= eta * x.grad
    x.grad.zero_()

# --- Create contour plot ---
x_vals = np.linspace(-12, 12, 100)
y_vals = np.linspace(-10, 10, 100)
Z = np.empty([100, 100])
for i in range(100):
    for j in range(100):
        Z[j, i] = f_float(x_vals[i], y_vals[j])  # note switched indices

plt.figure(figsize=(7,5))
contour_plot = plt.contour(x_vals, y_vals, Z, levels=15, cmap='viridis')
plt.colorbar(contour_plot, label='Function Value')

# Overlay gradient descent trajectory
plt.plot(trajectory_x1, trajectory_x2, marker='.', linestyle='-', color='red', label='Trajectory')

# Mark the global minimum
plt.plot(3, 2, marker='o', color='blue', markersize=8, label='Optimum')

plt.xlabel('x1')
plt.ylabel('x2')
plt.title(f'Gradient Descent Trajectory: η = {eta}')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Define the function
def f_np(x):
    return -np.exp(-(x - 1)**2) - 5*np.exp(-(x - 10)**2)

# Generate x-values
x_vals = np.linspace(-10, 20, 500)
y_vals = f_np(x_vals)

# Plot
plt.figure(figsize=(7,4))
plt.plot(x_vals, y_vals)
plt.title("f(x) = -exp(-(x-1)^2) - 5*exp(-(x-10)^2)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

The function shows two minima: one near x ≈ 1, and a deeper one near x ≈ 10.

This means:

If the starting point is near x = 1, it will likely converge to the local minimum at x ≈ 1 and miss the global minimum at x ≈ 10.

This is the "unavoidable problem of local minima."

 (b) Implement again the usual Gradient Decent algorithm using PyTorch as you have done before in
 order to find the minima of a function. Remember that your function has to be implemented using
 torch functions (in order for the auto-differentiation to work)!

```python
import torch

# Define the function in PyTorch
def f_torch(x):
    return -torch.exp(-(x-1)**2) - 5*torch.exp(-(x-10)**2)

# Gradient Descent function
def gradient_descent(x0, eta=0.1, n_steps=1000):
    x = torch.tensor(x0, requires_grad=True)
    trajectory_x = []
    trajectory_y = []

    for _ in range(n_steps):
        fx = f_torch(x)
        trajectory_x.append(float(x.detach()))
        trajectory_y.append(float(fx.detach()))

        fx.backward()
        with torch.no_grad():
            x -= eta * x.grad
        x.grad.zero_()

    return trajectory_x, trajectory_y, x
```

```python
import torch
import matplotlib.pyplot as plt
import numpy as np

# Constants
V = 355  # volume in cm³

# Surface area function as a function of radius using NumPy
def surface_area(r):
    return 2*V/r + 2*np.pi*r**2 # Unconstrained function formula

# PyTorch version for gradient descent
def surface_area_torch(r):
    return 2*V/r + 2*np.pi*r**2

# Gradient Descent
r = torch.tensor(3.0, dtype=torch.float32, requires_grad=True)  # initial guess;
# Typical beverage cans have a radius of around 3–4 cm
# Choosing r = 3 puts the algorithm in a reasonable range near the true minimum
eta = 0.01
n_steps = 1000

trajectory_r = []
trajectory_A = []

for n in range(n_steps):
    A = surface_area_torch(r)
    trajectory_r.append(float(r.detach()))
    trajectory_A.append(float(A.detach()))

    A.backward()
    with torch.no_grad():
        r -= eta * r.grad   # Gradient descent updates radius to minimize A(r)
    r.grad.zero_()

# Compute optimal height
r_opt = r.item()
h_opt = V / (np.pi * r_opt**2)
A_opt = 2*V/r_opt + 2*np.pi*r_opt**2

print(f"Optimal radius: {r_opt:.3f} cm")
print(f"Optimal height: {h_opt:.3f} cm")
print(f"Optimal surface area: {A_opt:.3f} cm²")
print(f"Material saved: {323 - A_opt:.3f} cm²")

# Plot surface area vs radius
r_vals = np.linspace(1, 10, 500)
A_vals = surface_area(r_vals)

plt.figure(figsize=(7,4))
plt.plot(r_vals, A_vals, label='Surface Area')
plt.scatter(trajectory_r, trajectory_A, s=12, color='red', label='Gradient Descent')
plt.xlabel('Radius r (cm)')
plt.ylabel('Surface Area A (cm²)')
plt.title('Optimization of Can Surface Area')
plt.grid(True)
plt.legend()
plt.show()
```
